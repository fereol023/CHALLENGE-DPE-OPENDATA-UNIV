```python
# Imports et configuration
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configuration de l'affichage
%matplotlib inline
plt.style.use('default')  # Utilisation du style par défaut
pd.set_option('display.max_columns', None)
```

```python
# Chargement des données
df = pd.read_parquet('../ressources/data/2_intermediary/enedis_ban_ademe_extract_PARIS_2022.parquet')

# Aperçu des données
print("Dimensions du dataset :", df.shape)
print("\nColonnes disponibles :")
print(df.columns.tolist())
print("\nAperçu des premières lignes :")
df.head()
```

Doublons :

```python
print("Nombre de doublons :", df.duplicated().sum())
df = df.drop_duplicates()
```

Valeurs manquantes :

```python
missing_values = df.isnull().sum()
missing_percentages = (missing_values / len(df)) * 100
missing_info = pd.DataFrame({
    'Valeurs manquantes': missing_values,
    'Pourcentage (%)': missing_percentages
})
print("\nColonnes avec valeurs manquantes :")
print(missing_info[missing_info['Valeurs manquantes'] > 0].sort_values('Pourcentage (%)', ascending=False))
```

```python
columns_to_drop = missing_info[missing_info['Pourcentage (%)'] > 90].index
df = df.drop(columns=columns_to_drop)
```

Types de variables

```python
df.info()
```

Traitement des objets 

```python
import numpy as np

for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].fillna(-1).infer_objects(copy=False)
    df[col] = df[col].str.replace(',', '.', regex=True)

    if "date" in col.lower():
        try:
            df[col] = pd.to_datetime(df[col], errors='coerce')
        except (ValueError, TypeError):
            print(f"Erreur sur {col} lors de la conversion en datetime.")

    else:
        try:
            df[col] = df[col].astype(np.int64)  # Conversion en entier
        except (ValueError, TypeError):
            try:
                df[col] = df[col].astype(np.float64)  # Conversion en flottant
            except (ValueError, TypeError):
                df[col] = df[col].astype("string")  # Seulement si int et float échouent
```

Tri des colonnes

```python
colonnes_ademe = [col for col in df.columns if "_ademe" in col]
colonnes_enedis_with_ban = [col for col in df.columns if "_enedis_with_ban" in col]
```
