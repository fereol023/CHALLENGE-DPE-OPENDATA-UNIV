```python
# Imports et configuration
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configuration de l'affichage
%matplotlib inline
plt.style.use('default')  # Utilisation du style par défaut
pd.set_option('display.max_columns', None)
```

```python
# Chargement des données
df = pd.read_parquet('../ressources/data/2_intermediary/enedis_ban_ademe_extract_PARIS_2022.parquet')

# Aperçu des données
print("Dimensions du dataset :", df.shape)
print("\nColonnes disponibles :")
print(df.columns.tolist())
print("\nAperçu des premières lignes :")
df.head()
```

Doublons :

```python
print("Nombre de doublons :", df.duplicated().sum())
df = df.drop_duplicates()
```

Valeurs manquantes :

```python
missing_values = df.isnull().sum()
missing_percentages = (missing_values / len(df)) * 100
missing_info = pd.DataFrame({
    'Valeurs manquantes': missing_values,
    'Pourcentage (%)': missing_percentages
})
print("\nColonnes avec valeurs manquantes :")
print(missing_info[missing_info['Valeurs manquantes'] > 0].sort_values('Pourcentage (%)', ascending=False))
```

```python
columns_to_drop = missing_info[missing_info['Pourcentage (%)'] > 90].index
df = df.drop(columns=columns_to_drop)
```

Types de variables

```python
df.info()
```

Traitement des objets 

```python
import numpy as np

for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].fillna(-1).infer_objects(copy=False)
    df[col] = df[col].str.replace(',', '.', regex=True)

    if "date" in col.lower():
        try:
            df[col] = pd.to_datetime(df[col], errors='coerce')
        except (ValueError, TypeError):
            print(f"Erreur sur {col} lors de la conversion en datetime.")

    else:
        try:
            df[col] = df[col].astype(np.int64)  # Conversion en entier
        except (ValueError, TypeError):
            try:
                df[col] = df[col].astype(np.float64)  # Conversion en flottant
            except (ValueError, TypeError):
                df[col] = df[col].astype("string")  # Seulement si int et float échouent
```

Tri des colonnes

```python
colonnes_ademe = [col for col in df.columns if "_ademe" in col]
colonnes_enedis_with_ban = [col for col in df.columns if "_enedis_with_ban" in col]
```

# Analyse exploratoire

Répartitiopn des DPE 

```python
ordre_dpe = ['A', 'B', 'C', 'D', 'E', 'F', 'G']

plt.figure(figsize=(10, 6))
sns.countplot(x='Etiquette_DPE_ademe', data=df, order=ordre_dpe)
plt.title("Répartition des Classes DPE")
plt.show()
```

```python
# 2. Consommation énergétique par type de chauffage
plt.figure(figsize=(12, 6))
sns.boxplot(x='Type_installation_chauffage_ademe', y='Conso_chauffage_é_finale_ademe', data=df)
plt.title("Consommation énergétique par type de chauffage")
plt.xticks(rotation=90)
plt.show()
```

```python
# 3. Répartition des étiquettes DPE par arrondissement (Code Postal)
plt.figure(figsize=(12, 6))
sns.countplot(x='district_enedis_with_ban', hue='Etiquette_DPE_ademe', data=df, palette='Set2')
plt.title("Répartition des étiquettes DPE par code postal")
plt.xlabel("Arrondissement Parisien")
plt.ylabel("Nombre de logements")
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Conso_5_usages_é_finale_ademe', y='Emission_GES_5_usages_ademe', data=df)
plt.title("Corrélation entre consommation énergétique et émission de GES")
plt.xlabel("Consommation énergétique finale")
plt.ylabel("Émission de GES")
plt.show()
```

```python
# 5. Évolution de la consommation énergétique par année de construction
plt.figure(figsize=(12, 6))
sns.barplot(x='Année_construction_ademe', y='Conso_5_usages_é_finale_ademe', data=df, ci=None)
plt.title("Consommation énergétique moyenne par année de construction")
plt.xticks(rotation=90)
plt.show()
```

```python
# 6. Impact de la qualité d'isolation sur la consommation
plt.figure(figsize=(10, 6))
sns.boxplot(x='Qualité_isolation_murs_ademe', y='Conso_chauffage_é_finale_ademe', data=df)
plt.title("Impact de l'isolation sur la consommation énergétique")
plt.xticks(rotation=90)
plt.show()
```

```python
# 7. Corrélation entre surface et consommation énergétique
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Surface_habitable_logement_ademe', y='Conso_5_usages_é_finale_ademe', data=df)
plt.title("Corrélation entre surface et consommation énergétique")
plt.xlabel("Surface habitable (m²)")
plt.ylabel("Consommation énergétique finale")
plt.show()
```
